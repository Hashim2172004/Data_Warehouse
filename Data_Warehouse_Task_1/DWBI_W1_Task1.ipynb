## 2. Decisions and assumptions

- **Scope:** all 38 rows are the stated morning's export. There are no timestamps, so the CSV alone cannot verify the morning cutoff. The five date strings are interpreted as **12 September 2026**, including the ambiguous `09-12-2026`, using the task/generator context rather than a global `dayfirst` guess.
- **Duplicate:** remove only the second exact copy of order 1004. It is Snacks, so this reduces all-category revenue by Rs 200 and Beverages revenue by **Rs 0**. Repeated IDs with different line details would require investigation, not automatic deletion.
- **Blank categories:** classify orders 1005 and 1033 as Dairy using the supplied catalogue. Their Rs 440 + Rs 880 = **Rs 1,320** remains outside Beverages. Effect on Beverages: **Rs 0**.
- **Return:** use net revenue and retain order 1007 at **−Rs 350**. It is an Electronics return, so its effect on Beverages is **Rs 0**. Do not turn the amount positive or discard it.
- **Amount error:** choose order 1001 as correct. It implies Rs 180 per cola bottle; orders 1002, 1003, 1021, 1030 and 1035 corroborate that price. The generator explicitly identifies 1008 as erroneous and gives a unit price of Rs 180. Correct 1008 from Rs 3,600 to **2 × Rs 180 = Rs 360**, decreasing Beverages by **Rs 3,240**.
- **Metric:** sum line `Amount`, which already includes quantity. Do not multiply it by `Qty` again. Treat amounts as PKR; no extra taxes or discounts are invented.


```python
# Parse amounts strictly after removing only the observed formatting.
work = raw.copy()
work['Amount_num'] = pd.to_numeric(
    work['Amount'].str.replace(r'^Rs\s*', '', regex=True).str.replace(',', '', regex=False),
    errors='raise'
)
work['Qty_num'] = pd.to_numeric(work['Qty'], errors='raise')
branch_map = {'LHR-Gulberg':'Lahore Gulberg', 'Lahore Gulberg':'Lahore Gulberg',
              'KHI-Clifton':'Karachi Clifton', 'khi clifton':'Karachi Clifton',
              'ISB-F7':'Islamabad F-7', 'ISB F-7':'Islamabad F-7'}
product_aliases = {'CL-150':'Cola 1.5L', 'Cola 1.5 Ltr':'Cola 1.5L',
                   'OJ-1L':'Orange Juice 1L', 'Milk 1 Litre':'Fresh Milk 1L'}
category_map = {'beverages':'Beverages', 'bev.':'Beverages', 'snacks':'Snacks',
                'dairy':'Dairy', 'grocery':'Grocery',
                'electrical':'Electronics', 'electronics':'Electronics'}
work['Branch_clean'] = work['Branch'].map(branch_map)
work['Product_clean'] = work['Product'].replace(product_aliases)
work['Category_clean'] = work['Category'].str.strip().str.lower().map(category_map)
# Explicit interpretations documented above; do not generalize this to other exports.
date_formats = {'12/09/2026':'%d/%m/%Y', '2026-09-12':'%Y-%m-%d',
                '09-12-2026':'%m-%d-%Y', '12 Sep 26':'%d %b %y', '2026/09/12':'%Y/%m/%d'}
work['Date_clean'] = work['Date'].map(lambda s: pd.to_datetime(s, format=date_formats[s]))

# Deduplicate using every original column, never just the order ID.
clean = work.loc[~raw.duplicated()].copy()
missing_category = clean['Category'].eq('')
clean.loc[missing_category, 'Category_clean'] = clean.loc[missing_category, 'Product_clean'].map({'Fresh Milk 1L':'Dairy'})
clean['Amount_clean'] = clean['Amount_num']
clean.loc[clean['OrderID'].eq('1008'), 'Amount_clean'] = 2 * 180

print('Cola price evidence before correction:')
cola = clean.loc[clean['Product_clean'].eq('Cola 1.5L'), ['OrderID','Qty_num','Amount_num']].copy()
cola['Implied_unit_price'] = cola['Amount_num'] / cola['Qty_num']
display(cola)
print(f"Clean rows: {len(clean)} | Branches: {clean['Branch_clean'].nunique()}")
```

## 3. Beverages revenue — calculation and reconciliation

Start with a deliberately incomplete exact-label total after safely parsing amounts. Add the sales labelled `Bev.`, then apply the documented decisions. This shows why getting a numeric sum alone is not enough.

```python
baseline = work.loc[work['Category'].eq('Beverages'), 'Amount_num'].sum()
alias_addition = work.loc[work['Category'].eq('Bev.'), 'Amount_num'].sum()
bridge = pd.DataFrame([
    ['Parsed amounts; exact Beverages label only', baseline],
    ['Include Bev. orders 1002, 1015, 1018, 1025', alias_addition],
    ['Remove duplicate 1004 (Snacks)', 0],
    ['Classify 1005 and 1033 as Dairy', 0],
    ['Retain return 1007 (Electronics)', 0],
    ['Correct order 1008: 360 minus 3,600', 360-3600],
], columns=['Step','Effect_PKR'])
bridge['Running_total_PKR'] = bridge['Effect_PKR'].cumsum()
display(bridge)

beverages = clean.loc[clean['Category_clean'].eq('Beverages')].copy()
display(beverages[['OrderID','Branch_clean','Product_clean','Qty_num','Amount','Amount_clean']])
revenue = beverages['Amount_clean'].sum()
print('Working:', ' + '.join(beverages['Amount_clean'].astype(str)), '=', revenue)
print(f'TOTAL NET BEVERAGES REVENUE: Rs {revenue:,.0f}')
print('\nBy product:')
display(beverages.groupby('Product_clean', as_index=False)['Amount_clean'].sum())
print('By branch:')
display(beverages.groupby('Branch_clean', as_index=False)['Amount_clean'].sum())
```

### Why amount parsing matters

`pd.to_numeric(..., errors='coerce')` on the raw field silently converts `Rs 180` and `1,280` to missing values. Both are Beverages orders. The following demonstration measures the resulting **Rs 1,460** understatement before correcting 1008.

```python
beverage_labels = work['Category'].isin(['Beverages', 'Bev.'])
naive = pd.to_numeric(raw['Amount'], errors='coerce')
print('Beverages with silently skipped amount strings:', naive.loc[beverage_labels].sum())
print('Beverages after safe amount parsing, before amount correction:', work.loc[beverage_labels, 'Amount_num'].sum())
print('Revenue silently skipped:', work.loc[beverage_labels, 'Amount_num'].sum() - naive.loc[beverage_labels].sum())

# Reconcile the result with independently stated catalogue prices.
expected_prices = {'Cola 1.5L':180, 'Orange Juice 1L':320}
independent_total = (beverages['Qty_num'] * beverages['Product_clean'].map(expected_prices)).sum()
assert revenue == independent_total == bridge['Running_total_PKR'].iloc[-1] == 6320
assert len(raw) == 38 and len(clean) == 37
assert clean['Branch_clean'].nunique() == 3
assert clean[['Branch_clean','Category_clean','Date_clean']].notna().all().all()
assert clean['Date_clean'].eq(pd.Timestamp('2026-09-12')).all()
assert clean.loc[clean['OrderID'].eq('1007'), 'Amount_clean'].iloc[0] == -350
assert clean['Amount_clean'].sum() == work['Amount_num'].sum() - 200 - 3240
assert raw.equals(original)
print('Checks passed: revenue reconciles, duplicate removed, return retained, raw data unchanged.')
```

## 5. Why not load it directly into a dashboard?

Loading this file unchanged would split identical branches and products, omit valid Beverages amounts, double-count a duplicate, and overstate one sale, producing a misleading dashboard even when its charts look correct.

## 6. Two problems preventable at data entry

| Problem | Example | One change to the branch system |
|---|---|---|
| Wrong line amount | Order 1008 records Rs 3,600 instead of Rs 360. | Make line amount a system-calculated field based on quantity and the approved product price, with any override requiring authorization. |
| Duplicate transaction | Order 1004 appears twice. | Enforce an idempotent transaction-line key so a repeated submission returns the original record instead of creating another one. |
